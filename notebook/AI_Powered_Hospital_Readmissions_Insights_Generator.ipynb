# **Exploratory Data Analysis**

```python
#Q1: Which readmission condition has the highest average excess readmission ratio
condition_summary = df_analysis.groupby('Measure Name').agg(
    Avg_Excess_Readmission_Ratio = ('Excess Readmission Ratio','mean'),
    Avg_Predicted_Readmission_Rate = ('Predicted Readmission Rate','mean'),
    Avg_Expected_Readmission_Rate = ('Expected Readmission Rate','mean'),
    Hospitals=("Facility ID", "nunique")
).sort_values('Avg_Excess_Readmission_Ratio', ascending=False)

condition_summary
```

Interpretation:
1. Across hospitals with reported data, hip/knee replacement had the highest average Excess Readmission Ratio, with predicted readmissions about 0.4% above the expected benchmark on average.
2. Heart Failure has the highest average predicted readmission rate, but its excess readmission ratio remains close to 1 because its expected rate is similarly high
3. CABG had one of the higher average Excess Readmission Ratios, but the result was based on fewer reporting hospitals than the other measures, so comparisons are interpreted with some caution.


```python
#Q2: Which state has the highest average excess readmission ratio
state_summary = df_analysis.groupby('State').agg(
    Avg_Excess_Readmission_Ratio = ('Excess Readmission Ratio','mean'),
    Avg_Predicted_Readmission_Rate = ('Predicted Readmission Rate','mean'),
    Avg_Expected_Readmission_Rate = ('Expected Readmission Rate','mean'),
    Hospitals=("Facility ID", "nunique")
).sort_values('Avg_Excess_Readmission_Ratio', ascending=False)

state_summary.head(10)
```

Interpretation:
1. Massachusetts had the highest average Excess Readmission Ratio among states in this analysis, with predicted readmissions approximately 3.4% above expected levels on average.

```python
#Q3: Which condition in MA state has the highest average excess readmission ratio
state_condition_summary = (
    df_analysis
    .groupby(["State", "Measure Name"])
    .agg(
        Avg_Excess_Readmission_Ratio=(
            "Excess Readmission Ratio", "mean"
        ),
        Avg_Predicted_Readmission_Rate=(
            "Predicted Readmission Rate", "mean"
        ),
        Avg_Expected_Readmission_Rate=(
            "Expected Readmission Rate", "mean"
        ),
        Hospitals=(
            "Facility ID", "nunique"
        )
    )
    .reset_index()
)

state_condition_summary[
    state_condition_summary["State"] == "MA"
].sort_values(
    "Avg_Excess_Readmission_Ratio",
    ascending=False
)
```

Interpretation:
Massachusetts had the highest average state-level Excess Readmission Ratio in the analyzed CMS data. AMI showed the largest relative gap, with predicted readmissions about 5.1% above expected levels, while several other conditions also exceeded their expected benchmarks, suggesting the state-level result was broad-based rather than attributable to a single condition.

# **Analysing additional information**

```python
# Do average Excess Readmission Ratios differ by hospital ownership type?
hospital_level_err = (
    df_hospital_analysis
    .groupby(
        ["Facility ID", "Hospital Ownership"],
        as_index=False
    )
    .agg(
        Hospital_Avg_ERR=(
            "Excess Readmission Ratio",
            "mean"
        )
    )
)
```

```python
ownership_summary = (
    hospital_level_err
    .groupby("Hospital Ownership")
    .agg(
        Avg_ERR=(
            "Hospital_Avg_ERR",
            "mean"
        ),
        Median_ERR=(
            "Hospital_Avg_ERR",
            "median"
        ),
        Hospitals=(
            "Facility ID",
            "nunique"
        )
    )
    .sort_values(
        "Avg_ERR",
        ascending=False
    ))
```

```python
ownership_summary
```

Interpretation: Hospital-level average ERR varied modestly by ownership type. Proprietary hospitals had the highest average among the larger ownership categories, while several nonprofit groups were slightly below the expected benchmark.


```python
# Does this ownership pattern still appear when we compare hospitals within the same clinical condition?
ownership_condition = (
    df_hospital_analysis
    .groupby(
        ["Hospital Ownership", "Measure Name"]
    )
    .agg(
        Avg_ERR=(
            "Excess Readmission Ratio",
            "mean"
        ),
        Hospitals=(
            "Facility ID",
            "nunique"
        )
    )
    .reset_index()
)

ownership_condition
```

```python
ownership_condition_pivot = (
    ownership_condition
    .pivot(
        index="Hospital Ownership",
        columns="Measure Name",
        values="Avg_ERR"
    )
)

ownership_condition_pivot
```

Interpretation: Proprietary hospitals showed average Excess Readmission Ratios above the expected benchmark across all six HRRP conditions in this dataset. The largest relative gaps were observed for CABG and hip/knee replacement. This indicates a consistent association between proprietary ownership and higher average ERR across conditions, but it does not establish that ownership causes higher readmissions.

```python
ownership_condition_count = (
    ownership_condition
    .pivot(
        index="Hospital Ownership",
        columns="Measure Name",
        values="Hospitals"
    )
)

ownership_condition_count
```

Interpretation: Proprietary hospitals showed average ERR above 1.0 across all six HRRP measures, with each condition represented by more than 100 reporting hospitals. The association was therefore broad-based across conditions rather than driven by a single measure or very small sample.

# **Key project Takaways**

- HIP-KNEE had the highest average Excess Readmission Ratio across the six HRRP conditions, although the average was only about 0.4% above the expected benchmark.

- Massachusetts had the highest average state-level ERR among the states analyzed, at approximately 3.4% above expected.
